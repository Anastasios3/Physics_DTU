## Exercise 1: Velocity selector



Book: University Physics Vol. 2.

1. Acceleration: in a uniform field the work over a distance $L$ is $W = eE_aL$ (p. 281), and all of it becomes kinetic energy, as in Example 7.6 (p. 279): $eE_aL = \tfrac12 m_e v^2$.
2. Straight line: the electric and magnetic forces cancel, $eE = evB$, Eq. 11.24 (p. 482). This is the velocity selector, Eq. 11.25 (p. 482) and Example 11.8 (p. 483): $v = E/B$, so $B = E/v$.

```python
e = 1.602e-19       # C
me = 9.11e-31       # kg
Ea = 2.5e2          # V/m
L = 0.73            # m
E = 3.5e4           # V/m

v = (2 * e * Ea * L / me) ** 0.5
B = E / v

print(f"a) v = {v:.3g} m/s")
print(f"b) B = {B:.3g} T")
```

a) $v \approx 8.01\cdot10^{6}$ m/s

b) $B \approx 4.37\cdot10^{-3}$ T $= 4.37$ mT

The forces only cancel if the electron moves up the page: $-e\vec E$ pushes it left and $-e\vec v\times\vec B$ pushes it right.

## Exercise 2: Hall sensor



Book: University Physics Vol. 2, Section 11.6 The Hall Effect (p. 481).

$x$ to the right, $y$ up, $z$ out of the page. Then $\vec B = -B\,\hat z$, and the positive carriers move with the current, $\vec v_d = v_d\,\hat y$.

1. Magnetic force, Eq. 11.1 (p. 466): $\vec F = e\vec v_d\times\vec B = ev_dB\,(\hat y\times(-\hat z)) = -ev_dB\,\hat x$.
2. Equilibrium, Eq. 11.24 (p. 482): $eE = ev_dB$, with the electric force opposite to the magnetic force.
3. Hall voltage, Eq. 11.29 (p. 483): $V = IBl/(neA)$. Here $l = w$ (the distance from A to B) and $A = wh$ (the cross section the current flows through), so $w$ cancels: $V = IB/(neh)$, with $h$ the thickness along $\vec B$.

```python
e = 1.602e-19       # C
I = 10.0e-3         # A
B = 1.0e-4          # T
n = 3.5e21          # m^-3
h = 0.5e-2          # m, thickness along B

V = I * B / (n * e * h)
print(f"|VB - VA| = {V:.3g} V")
```

a) To the left, toward the A side: $\vec F = -ev_dB\,\hat x$.

b) The positive carriers pile up on the left edge, so the transverse field points to the right, from A toward B. A is at the higher potential, so $V_B - V_A$ is negative.

c) $V_B - V_A = -\dfrac{IB}{neh} \approx -3.57\cdot10^{-7}$ V $\approx -0.357$ µV

$w$ drops out: a wider strip gives a larger $V = Ew$ but a proportionally smaller drift speed $v_d = I/(newh)$.

## Exercise 3: Three parallel wires



Book: University Physics Vol. 2, Section 12.3, Eq. 12.11 (p. 509): $F/l = \mu_0 I_1 I_2/(2\pi r)$, attractive when the currents point the same way.

All three currents point into the page and all distances are equal, so the top wire is pulled toward each bottom wire with the same force. Add the two forces as vectors, in units of $\mu_0 I^2/(2\pi a)$ with side $a = 1$.

```python
import math

top = (0.5, math.sqrt(3) / 2)
others = [(0, 0), (1, 0)]       # bottom left, bottom right

Fx, Fy = 0, 0
for x, y in others:
    rx, ry = x - top[0], y - top[1]     # from the top wire toward the other wire
    r = math.hypot(rx, ry)
    Fx += rx / r                        # attraction, size 1 each
    Fy += ry / r

print("Fx =", round(Fx, 3))
print("Fy =", round(Fy, 3))
```

Answer: B) Downwards

The sideways pulls cancel by symmetry and the two downward parts add: $F/l = \sqrt3\,\mu_0 I^2/(2\pi a)$.

## Exercise 4: Electron moving next to a wire



Book: University Physics Vol. 2.

1. Field of a long straight wire, Eq. 12.8 (p. 506): $B = \mu_0 I/(2\pi r)$. Direction by RHR-2 (p. 506): thumb along $I$ (up), so on the right of the wire $\vec B$ points into the page.
2. Force, Eq. 11.1 (p. 466): $\vec F = q\vec v\times\vec B$ with $q = -e$. $\vec v \perp \vec B$, so $F = evB$ (Eq. 11.2).

$x$ right, $y$ up, $z$ out of the page: $\vec v = v\,\hat y$ and $\vec B = -B\,\hat z$, so $\vec F = -e\,vB\,(\hat y\times(-\hat z)) = +evB\,\hat x$, to the right.

```python
import math

mu0 = 4 * math.pi * 1e-7    # T m/A
e = 1.602e-19               # C
I = 2.7                     # A
v = 3.5e6                   # m/s
r = 0.12                    # m

B = mu0 * I / (2 * math.pi * r)
F = e * v * B
print(f"B = {B:.3g} T")
print(f"F = {F:.3g} N")
```

Answer: B) $2.5\cdot10^{-18}$ N, towards the right

Hand check: an electron moving up is a current pointing down, opposite to $I$, and opposite currents repel (p. 509). Forgetting the minus sign of the electron gives F) (left).

## Exercise 5: Electron circling above a wire



Book: University Physics Vol. 2.

1. Frequency: the period of the circular motion is Eq. 11.6 (p. 471), $T = 2\pi m/(qB)$, so $f = 1/T = eB/(2\pi m_e)$ and $B = 2\pi m_e f/e$. It does not depend on $r$ or $v$.
2. This $B$ comes from the wire, Eq. 12.8 (p. 506): $B = \mu_0 i/(2\pi d)$, so $i = 2\pi d B/\mu_0$.
3. Direction: the arrow on the circle points down on its left side, so the electron goes counterclockwise. Figure 11.7 (p. 471) shows a negative charge going clockwise when $\vec B$ points into the page, so here $\vec B$ points out of the page. Above the wire, RHR-2 (p. 506) gives $\vec B$ out of the page when $i$ points to the right.

```python
import math

mu0 = 4 * math.pi * 1e-7    # T m/A
e = 1.602e-19               # C
me = 9.11e-31               # kg
f = 1.0e7                   # 1/s
d = 0.010                   # m

B = 2 * math.pi * me * f / e
i = 2 * math.pi * d * B / mu0
print(f"B = {B:.3g} T")
print(f"i = {i:.3g} A")
```

Answer: F) $i = 17.9$ A, towards the right

Using $\omega = eB/m$ as if it were $f$ loses a factor $2\pi$ and gives 2.8 A (options A and B).

## Exercise 6: Torque on an atom between two wires



Book: University Physics Vol. 2.

$x$ right, $y$ up, $z$ out of the page, as in the figure. Both currents point up ($+y$).

1. Field of each wire, Eq. 12.8 (p. 506): $B = \mu_0 I/(2\pi r)$ with $r = d/2$. RHR-2 (p. 506): the midpoint is right of wire 1, so $\vec B_1$ points into the page ($-z$); it is left of wire 2, so $\vec B_2$ points out of the page ($+z$). Net: $B_z = \mu_0 (I_2 - I_1)/(\pi d)$.
2. Torque, Eq. 11.22 (p. 481): $\vec\tau = \vec\mu\times\vec B$ with $\vec\mu = \mu_B(\cos45^\circ,\ \sin45^\circ,\ 0)$. With $\vec B$ along $z$ this gives $\tau_x = \mu_y B_z$.

```python
import math
import numpy as np

mu0 = 4 * math.pi * 1e-7    # T m/A
muB = 9.27e-24              # J/T
I1, I2 = 3.00, 10.0         # A, both along +y
d = 0.100                   # m
r = d / 2                   # m, midpoint to each wire

B1z = -mu0 * I1 / (2 * math.pi * r)    # right of wire 1: into the page
B2z = +mu0 * I2 / (2 * math.pi * r)    # left of wire 2: out of the page

a = math.radians(45)
mu = muB * np.array([math.cos(a), math.sin(a), 0])
B = np.array([0, 0, B1z + B2z])
tau = np.cross(mu, B)

print(f"Bz = {B[2]:.3g} T")
print(f"tau_x = {tau[0]:.3g} N m")
```

Answer: A) $\tau_x = 1.84\cdot10^{-28}$ N m

$I_2 > I_1$, so the net field points out of the page. A sign slip in the right-hand rule flips $\vec B$ and gives D) $-1.84\cdot10^{-28}$ N m. Adding the two fields instead of subtracting gives C) $3.41\cdot10^{-28}$ N m.
