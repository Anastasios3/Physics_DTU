## Exercise 1: Proton in a magnetic field



Book: University Physics Vol. 2, Sections 11.2 and 11.3.

1. Force on a moving charge, Eq. 11.1 (p. 466): $\vec F = q\vec v\times\vec B$. Direction by the right-hand rule RHR-1 (p. 466 to 467): sweep the fingers from $\vec v$ to $\vec B$, the thumb gives $\vec F$. The acceleration $\vec a = \vec F/m$ points the same way.
2. $\vec F$ is always perpendicular to $\vec v$, so the magnetic force does no work and the kinetic energy stays constant (p. 470).

Axes: $x$ to the right, $y$ up, $z$ out of the page. Only directions are asked, so all sizes are 1.

```python
import numpy as np

q = +1                      # proton, in units of e
v = np.array([1, 0, 0])     # to the right
B = np.array([0, 0, -1])    # into the page

F = q * np.cross(v, B)
print("F direction:", F)
print("F . v =", np.dot(F, v))
```

$\vec F$ points along $+y$, and $\vec F\cdot\vec v = 0$.

Answer: A) and H) are correct.

- A: fingers to the right, curl into the page, thumb up. $\vec v \perp \vec B$, so $F = evB \neq 0$ (not E), and $\vec F \perp \vec B$ keeps it in the page (not C or D).
- H: the force is perpendicular to the velocity, so it only turns the proton. Speed and kinetic energy stay the same (not F or G).

## Exercise 2: Protons in a synchrotron



Book: University Physics Vol. 2.

1. Kinetic energy in joules: $1\ \text{eV} = 1.602\cdot10^{-19}$ J (p. 278). Speed from $K = \tfrac12 m v^2$.
2. Radius of the circle, Eq. 11.5 (p. 471): $r = mv/(qB)$, so $B = mv/(er)$.

```python
import math

e = 1.602e-19       # C
m = 1.67e-27        # kg, proton
K = 2.0e6 * e       # J
r = 0.50            # m
c = 3.00e8          # m/s

v = math.sqrt(2 * K / m)
B = m * v / (e * r)

print(f"v = {v:.3g} m/s = {v / c:.3f} c")
print(f"B = {B:.3g} T")
```

Answer: C) $B \approx 0.41$ T

$v \approx 0.07\,c$, so $K = \tfrac12 mv^2$ (no relativity) is fine. Dropping the $\tfrac12$ in $K$ gives B) 0.29 T, and using the diameter 1.0 m as the radius gives A) 0.20 T.

## Exercise 3: Rectangular current loop



Book: University Physics Vol. 2, Sections 11.4 and 11.5.

1. Force on a straight wire, Eq. 11.13 (p. 476): $\vec F = I\vec l\times\vec B$, direction by RHR-1 with the fingers along the current (p. 476). Every side is perpendicular to $\vec B$, so $F = IlB$.
2. Magnetic moment, Eq. 11.20 (p. 480): $\vec\mu = IA\,\hat n$ with $A = ab$. RHR-2 (p. 480): curl the fingers along the current, the thumb gives $\hat n$.
3. Torque, Eq. 11.22, and energy, Eq. 11.23 (p. 481): $\vec\tau = \vec\mu\times\vec B$ and $U = -\vec\mu\cdot\vec B$. Turned slowly, the work done on the loop is $W = U_f - U_i$.

```python
I = 10.0            # A
B = 0.154           # T
a = 0.200           # m, width
b = 0.100           # m, height

F_ab = I * a * B    # top and bottom sides
F_b = I * b * B     # left and right sides
mu = I * a * b

U_i = -mu * B       # mu parallel to B
U_f = +mu * B       # after 180 degrees: mu antiparallel to B
W = U_f - U_i

print(f"a) top/bottom: F = {F_ab:.3g} N,  left/right: F = {F_b:.3g} N")
print(f"b) mu = {mu:.3f} A m^2")
print(f"d) U_i = {U_i:.3g} J,  U_f = {U_f:.3g} J,  W = {W:.3g} J")
```

a) Top side: 0.308 N up. Bottom side: 0.308 N down. Right side: 0.154 N to the right. Left side: 0.154 N to the left. All four forces point outward and cancel, so the net force is zero (Eq. 11.17, p. 480).

b) $\mu = 0.200$ A m², into the page (the current runs clockwise), so $\vec\mu$ is parallel to $\vec B$.

c) $\tau = 0$. Each force acts at the middle of its side and points straight away from the center, so its lever arm is zero. Eq. 11.22 agrees: $\vec\mu \parallel \vec B$ gives $\vec\mu\times\vec B = 0$.

d) $W = 2\mu B \approx 0.0616$ J. $U$ goes from $-\mu B = -0.0308$ J (lowest possible) to $+\mu B = +0.0308$ J (highest possible).

Check for the top side: fingers to the right, curl into the page, thumb up.

## Exercise 4: Electron through an E-field into a B-field

c) Clockwise. The electron enters moving to the right with $\vec B$ into the page. $\vec v\times\vec B$ points up, and the negative charge flips it, so the force points down. The electron bends downward, which is clockwise, as in Figure 11.7.

## Exercise 5: Triangular loop with B in its plane



Book: University Physics Vol. 2, Section 11.5.

1. $\vec\mu = IA\,\hat n$, Eq. 11.20 (p. 480), with the area of an equilateral triangle $A = \tfrac{\sqrt3}{4}a^2$. The current runs clockwise, so by RHR-2 (p. 480) $\vec\mu$ points into the page.
2. $\vec\tau = \vec\mu\times\vec B$, Eq. 11.22 (p. 481). $\vec\mu \perp \vec B$, so $\tau = \mu B\sin 90^\circ = IAB$.
3. $U = -\vec\mu\cdot\vec B$, Eq. 11.23 (p. 481): now $U = 0$ ($90^\circ$), and the minimum is $-\mu B$ ($\vec\mu \parallel \vec B$). Work done on the loop: $W = U_f - U_i$.

```python
import math

I = 5.00            # A
a = 0.300           # m
B = 1.20            # T

A = math.sqrt(3) / 4 * a**2
mu = I * A
tau = mu * B        # sin 90 = 1
W = -mu * B - 0     # U_min - U_now

print(f"A = {A:#.3g} m^2,  mu = {mu:.3g} A m^2")
print(f"a) tau = {tau:.3g} N m")
print(f"b) W = {W:.3g} J")
```

a) $\tau \approx 0.234$ N m. $\vec\tau = \vec\mu\times\vec B$ lies in the page, pointing straight down (perpendicular to $\vec B$). The bottom side is parallel to $\vec B$ and feels no force. The left side is pushed into the page and the right side out of it, so the loop turns about a vertical axis until $\vec\mu$ points along $\vec B$.

b) $W = -\mu B \approx -0.234$ J. The work is negative because the field turns the loop by itself: to turn it slowly you have to hold it back.

Hand check of a): each slanted side feels $F = IaB\sin 60^\circ$ at a lever arm $a/4$, so $\tau = 2\cdot IaB\tfrac{\sqrt3}{2}\cdot\tfrac a4 = IB\tfrac{\sqrt3}{4}a^2$.

## Exercise 6: Electron circling a positive charge



Book: University Physics Vol. 2.

1. Electric force: $Q$ is spherically symmetric, so it acts like a point charge at its center (p. 285). Coulomb's law, Eq. 5.1 (p. 179): $F_E = keQ/R^2$, attractive, so toward $Q$.
2. Magnetic force, Eq. 11.2 (p. 466) with $\theta = 90^\circ$: $F_B = evB$. Direction by RHR-1, reversed for a negative charge (step 4, p. 467).
3. Newton's 2nd law for uniform circular motion: the net force toward the center equals $mv^2/R$, as in Eq. 11.4 (p. 471).

In the figure the electron is at the top moving to the right, so it circles clockwise.

```python
k = 8.99e9          # N m^2/C^2
e = 1.602e-19       # C
me = 9.11e-31       # kg
Q = 0.1e-9          # C
v = 1e6             # m/s
R = 0.04            # m

F_E = k * e * Q / R**2
F_c = me * v**2 / R         # net force needed toward the center
F_B = F_E - F_c             # what the magnetic force must take away
B = F_B / (e * v)

print(f"a) F_E = {F_E:.2e} N,  needed m v^2/R = {F_c:.2e} N")
print(f"d) F_B = {F_B:.2e} N,  B = {B:.2e} T")
```

a) $F_E = \dfrac{keQ}{R^2} \approx 9.00\cdot10^{-17}$ N toward $Q$, and $F_B = evB$.

b) At the top, $\vec v$ points right. $\vec B$ out of the page: $\vec v\times\vec B$ points down, flipped for the electron, so $F_B$ points away from $Q$. $\vec B$ into the page: $F_B$ points toward $Q$.

c) Toward the center as positive, with $\vec B$ out of the page:
$$\frac{mv^2}{R} = \frac{keQ}{R^2} - evB$$
(with $\vec B$ into the page the sign of $evB$ is $+$).

d) $$B = \frac{keQ/R^2 - mv^2/R}{ev} \approx 4.20\cdot10^{-4}\ \text{T, out of the page}$$

The Coulomb pull ($9.00\cdot10^{-17}$ N) is larger than the $2.28\cdot10^{-17}$ N needed for this circle, so the magnetic force must push outward. With $\vec B$ into the page both forces pull inward, and $B$ would come out negative.
