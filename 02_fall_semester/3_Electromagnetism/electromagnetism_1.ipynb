## Exercise 1: Four charges in a square



Book: University Physics Vol. 2, Section 5.4 Electric Field.

Equation 5.4 (p. 185, PDF page 197):

$$\vec E(P) = \frac{1}{4\pi\varepsilon_0}\sum_{i=1}^{N}\frac{q_i}{r_i^2}\,\hat r_i$$

$\vec r_i$ goes from charge $i$ to the point $P$ and $\hat r_i = \vec r_i / r_i$, so each term is $k\,q_i\,\vec r_i / r_i^3$ with $k = 1/(4\pi\varepsilon_0)$.
Keep the sign of $q_i$ and the direction comes out by itself: away from $+q$, toward $-q$ (rule on p. 186).

Square with side $a$, bottom left corner at the origin, dot at $P = (a,\ a/2)$.
No values are given and only the direction is asked, so $k = q = a = 1$. The field comes out in units of $kq/a^2$.

```python
k, q, a = 1, 1, 1

charges = [(+q, 0, a),   # top left
           (-q, a, a),   # top right
           (-q, 0, 0),   # bottom left
           (+q, a, 0)]   # bottom right

Px, Py = a, a / 2        # the black dot

Ex, Ey = 0, 0
for qi, x, y in charges:
    rx, ry = Px - x, Py - y
    r = (rx**2 + ry**2) ** 0.5
    Ex += k * qi * rx / r**3
    Ey += k * qi * ry / r**3

print("Ex =", round(Ex, 3))
print("Ey =", round(Ey, 3))
```

$E_x = 0$ and $E_y > 0$, so the total field points straight up.

Answer: D) Upwards

Hand check with the rule on p. 186:

1. Right pair: the dot sits between them, $a/2$ from each. Away from $+q$ (below) and toward $-q$ (above) both point up.
2. Left pair: same setup as Example 5.4(b), Equation 5.6 (p. 189, PDF page 201), with $z = d = a$. The horizontal parts cancel and the vertical parts point from $+q$ toward $-q$, so down.
3. Each left charge is $\sqrt5$ times farther away ($a\sqrt5/2$ against $a/2$), so its field is 5 times weaker, and only part of it points down. Up wins.

## Exercise 2: Electron in static equilibrium



Book: University Physics Vol. 2.

The two charge distributions are spherically symmetric, so each acts like a point charge at its center (Section 6.3, Eq. 6.9, p. 237). Coulomb's law is Eq. 5.1 (p. 179), and the net force is the sum of the forces, Eq. 5.2 (p. 181).

Both charges are to the right of the electron. A charge $Q$ at distance $r$ on that side gives the electron $F_x = k e Q / r^2$: to the right if $Q > 0$ (attraction), to the left if $Q < 0$ (repulsion).

Static equilibrium: $k e\,(q_1/r_1^2 + q_2/r_2^2) = 0$, so $k$ and $e$ cancel. Distances from the electron: $r_1 = 1$ cm, $r_2 = 1 + 2 = 3$ cm.

```python
q1 = 1e-9                     # C
r1, r2 = 0.01, 0.01 + 0.02    # m, from the electron

q2 = -q1 * (r2 / r1)**2
print("q2 =", round(q2 * 1e9, 2), "nC")
```

Answer: F) $q_2 = -9$ nC

$q_1$ pulls the electron to the right, so $q_2$ must push it to the left, which takes a negative charge. It is 3 times farther away, so it needs 9 times the charge.

## Exercise 3: Braking a deuterium ion



Book: University Physics Vol. 2.

1. Direction: $\vec F = q\vec E$, Eq. 5.3 (p. 184). The ion is positive, so the force points along $\vec E$. To slow it down, $\vec E$ must point against $\vec v$.
2. Size: in a uniform field the work over a distance $d$ is $W = qEd$ (p. 281). By conservation of energy (p. 279) this work removes all the kinetic energy: $eEd = \tfrac12 m v^2$.

A deuterium ion is a proton plus a neutron, so $m = 2 \times 1.67\cdot10^{-27}$ kg and $q = +e$.

```python
m = 2 * 1.67e-27    # kg
e = 1.602e-19       # C
v = 1e5             # m/s
d = 0.10            # m

E = m * v**2 / (2 * e * d)
print("E =", round(E), "V/m")
```

Answer: D) $|\vec E| \approx 1042$ V/m, opposite to $\vec v$

With one nucleon mass instead of two you get 521 V/m (option E).

## Exercise 4: Electron released next to two charges



Book: University Physics Vol. 2.

1. Spherically symmetric charges act like point charges at their centers (p. 285).
2. Potential from several point charges, Eq. 7.9 (p. 288): $V = k\,(q_1/r_1 + q_2/r_2)$. Potential energy of the electron: $U = qV = -eV$ (p. 288). The $q_1 q_2$ pair is held fixed, so its energy never changes and is left out.
3. Conservation of energy (p. 279): $K_i + U_i = K_f + U_f$. The electron starts at rest ($K_i = 0$) and ends far away ($U_f = 0$), so $\tfrac12 m_e v^2 = U_i$.

Distances from the electron: $r_1 = 0.30$ m and $r_2 = 0.30 + 1.0 = 1.30$ m.

```python
k = 8.99e9                    # N m^2/C^2
e = 1.602e-19                 # C
me = 9.11e-31                 # kg
q1, q2 = 3e-9, -70e-9         # C
r1, r2 = 0.30, 0.30 + 1.0     # m, from the electron

V = k * (q1 / r1 + q2 / r2)
U = -e * V
v = (2 * U / me) ** 0.5

print(f"a) U = {U:.3g} J  ({U / e:.0f} eV)")
print(f"b) v = {v:.3g} m/s")
```

a) $U \approx 6.3\cdot10^{-17}$ J. It is positive because the large negative $q_2$ makes $V < 0$ where the electron sits.

b) $v \approx 1.2\cdot10^{7}$ m/s

## Exercise 5: Field strength at A



Same method as Exercise 1: Eq. 5.4 (p. 185) summed over the four charges, now with numbers. Bottom left corner at the origin, A at $(0,\ s/2)$, $k = 8.99\cdot10^9$ N m²/C² (p. 180).

```python
k = 8.99e9          # N m^2/C^2
q1 = 100e-9         # C
s = 0.50            # m

charges = [(-q1, 0, s),   # top left
           (+q1, s, s),   # top right
           (+q1, 0, 0),   # bottom left
           (-q1, s, 0)]   # bottom right

Px, Py = 0, s / 2         # point A

Ex, Ey = 0, 0
for qi, x, y in charges:
    rx, ry = Px - x, Py - y
    r = (rx**2 + ry**2) ** 0.5
    Ex += k * qi * rx / r**3
    Ey += k * qi * ry / r**3

E = (Ex**2 + Ey**2) ** 0.5
print("Ex =", round(Ex), "N/C")
print("Ey =", round(Ey), "N/C")
print("E  =", round(E), "N/C")
```

Answer: E) $E = 2.6\cdot10^4$ N/C

Same picture as Exercise 1: the two charges on the left side both give a field pointing up, and the far pair takes away only a small part.

## Exercise 6: Electron under a charged plane



Book: University Physics Vol. 2 for the field, Vol. 1 for projectile motion.

1. Field of an infinite plane, Eq. 5.15 (p. 197): $E = \sigma/(2\varepsilon_0)$, the same at every distance. It points toward a negative plane (rule on p. 186).
2. Force, Eq. 5.3 (p. 184): $\vec F = -e\vec E$, so the electron is pushed away from the plane, down toward the material. The force is constant, so the acceleration $a = eE/m_e$ is constant.
3. Constant acceleration straight down and none along $x$ is projectile motion with $g$ replaced by $a$ (Vol. 1, Section 4.3): maximum height $h = v_{0y}^2/(2a)$ (p. 162), range $R = v_0^2 \sin 2\theta / a$, Eq. 4.26 (p. 165).

$m_e$ and $e$ as in Exercise 4, $\varepsilon_0 = 8.85\cdot10^{-12}$ C²/(N m²) (p. 180).

```python
import math

e = 1.602e-19       # C
me = 9.11e-31       # kg
eps0 = 8.85e-12     # C^2/(N m^2)
sigma = 1.0e-6      # C/m^2
d = 0.010           # m
v0 = 2.0e7          # m/s
theta = math.radians(60)

E = sigma / (2 * eps0)
F = e * E
a = F / me

h = (v0 * math.sin(theta))**2 / (2 * a)
R = v0**2 * math.sin(2 * theta) / a

print(f"a) E = {E:.3g} N/C,  F = {F:.3g} N,  a = {a:.3g} m/s^2")
print(f"c) h = {h * 100:.2f} cm,  d = {d * 100:.2f} cm,  R = {R * 100:.2f} cm")
```

a) $E \approx 5.6\cdot10^4$ N/C, pointing up toward the plane. $F \approx 9.1\cdot10^{-15}$ N on the electron, pointing down, away from the plane.

b) A parabola, Eq. 4.25 (p. 165): projectile motion with $g$ replaced by $a$.

c) No. The electron would rise $h \approx 1.5$ cm before turning, but the plane is only $d = 1.0$ cm away. It reaches the charged plane on the way up and never falls back. $R \approx 3.5$ cm would be the landing point only if $h < d$.

## Exercise 7: Two deuterium ions



Book: University Physics Vol. 2.

1. Potential energy of two point charges, Eq. 7.2 (p. 271): $U = k q_1 q_2 / r$, zero when they are infinitely far apart.
2. Conservation of energy (p. 279): $K_i + U_i = K_f + U_f$.

At the start both ions move with speed $v$ and $U_i = 0$. Their momenta are equal and opposite, so at the closest point both stand still: $K_f = 0$ and $U_f = k e^2/r$. That gives $2 \cdot \tfrac12 m v^2 = k e^2 / r$ with $m = 2 \times 1.67\cdot10^{-27}$ kg (proton plus neutron).

```python
k = 8.99e9          # N m^2/C^2
e = 1.602e-19       # C
m = 2 * 1.67e-27    # kg
r = 1e-15           # m

v = (k * e**2 / (m * r)) ** 0.5
print(f"v = {v:.3g} m/s")
```

Answer: D) $|\vec v| \approx 8.31\cdot10^6$ m/s

Option E is $v^2$, the number before the square root.

## Exercise 8: Triple alpha process



Book: University Physics Vol. 2.

1. Each alpha particle is spherically symmetric, so it acts like a point charge $2e$ at its center (p. 285). When they just touch, the centers are $2R$ apart.
2. The energy of a group of charges is the sum over every pair, Eq. 7.3 (p. 274), as in Example 7.3 (p. 272). Three particles make three pairs.
3. Conservation of energy (p. 279): far apart $U_i = 0$. The smallest $K_i$ is when all of it has turned into $U$ at the moment they touch ($K_f = 0$), so $K_i = U_f$.

```python
k = 8.99e9          # N m^2/C^2
e = 1.602e-19       # C
q = 2 * e           # C
R = 1.68e-15        # m
r = 2 * R           # m, center to center

Ki = 3 * k * q * q / r
print(f"Ki = {Ki:.3g} J = {Ki * 1e12:.3f} pJ")
```

Answer: D) 0.824 pJ

Using $R$ instead of $2R$ gives E) 1.65 pJ, and using $e$ instead of $2e$ gives A) 0.206 pJ.
