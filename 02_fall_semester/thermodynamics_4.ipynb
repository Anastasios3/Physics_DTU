```python
R = 8.314
n = 1.0
T1 = 273.15

T2 = 3 * T1

dT = T2 - T1
Cp = 2.5 * R
Q = n * Cp * dT

print("T1 =", T1, "K")
print("T2 =", T2, "K")
print("dT =", dT, "K")
print("Cp =", round(Cp, 3), "J/(mol K)")
print("Q  =", round(Q, 1), "J  =", round(Q / 1000, 2), "kJ")
```

```python
import math

R = 8.314
n = 1.0
ratio = 1 / 5

dS = n * R * math.log(ratio)

print("ln(V2/V1) =", round(math.log(ratio), 4))
print("dS =", round(dS, 2), "J/K per mole")
```

```python


def entropies(Q, Th, Tc):
    dS_rod = 0.0
    dS_hot = -Q / Th
    dS_cold = Q / Tc
    dS_universe = dS_rod + dS_hot + dS_cold
    return dS_rod, dS_hot, dS_cold, dS_universe

rod, hot, cold, universe = entropies(100.0, 500.0, 300.0)

print("dS_rod=", round(rod, 4), "J/K")
print("dS_hot=", round(hot, 4), "J/K")
print("dS_cold=", round(cold, 4), "J/K")
print("dS_universe=", round(universe, 4), "J/K  (positive, so the process is allowed)")
```

```python
import math

m = 0.200
c = 4186.0
T1 = 0.0  + 273.15
T2 = 80.0 + 273.15

Q = m * c * (T2 - T1)

dS_water = m * c * math.log(T2 / T1)
dS_res = -Q / T2
dS_universe = dS_water + dS_res

print("final water temp= 80.0 C")
print("final reservoir temp= 80.0 C  (unchanged)")
print("Q  =", round(Q, 1), "J  =", round(Q / 1000, 2), "kJ")
print("dS_water=", round(dS_water, 2), "J/K")
print("dS_reservoir=", round(dS_res, 2), "J/K")
print("dS_universe=", round(dS_universe, 2), "J/K")
```
