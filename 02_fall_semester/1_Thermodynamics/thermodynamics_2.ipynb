```python
p1 = 1.00
T1 = 22.0 + 273.15
T2 = 32.0 + 273.15
ratio_V = 1.38

p2 = p1 * (1 / ratio_V) * (T2 / T1)

print(p2, "atm")
print(p2 * 101.325, "kPa")
```

```python
R = 8.314
n = 0.500
T = 25.0 + 273.15
V = 2.00e-3
M = 28.0e-3
g = 9.80
p = n * R * T / V
a = V ** (1/3)
A = a * a
F = p * A
m = n * M
W = m * g

print(p, "Pa")
print(F, "N")
print(W, "N")
print(F / W)
```

```python
R = 8.314
p1 = 101325
V = 0.500e-3
T1 = 80.0 + 273.15
T2 = 25.0 + 273.15

n = p1 * V / ( R * T1 )
p2 = p1 * T2 / T1
pg = p2 - p1
print(n, "mol")
print(p2, "Pa")
print(pg, "Pa")
```
